# 线性表

定义

```c++17
线性表是包含若干元素的一个线性序列
记为:L = (a0, a1, ..., a(n-1))
其中 L是表名，n是表长，ai (0<=i<=n-1)是表中的元素
             n > 0 时称为非空表，n = 0 时称为空表。

L = (D,R)  
D是数据元素的集合 R是数据元素之间的关系集合

/*从纯理论的数据结构角度来看，一个线性表（如数组、链表）通常被定义为一组相同类型的数据元素的有限序列。这种“类型相同”的约束是为了：
操作一致性：确保对每个元素进行的操作（如比较、排序、计算）都是有效的。
内存管理：相同类型的元素占用相同大小的内存空间，使得通过索引进行常数时间访问（例如在数组中）成为可能。
类型安全：避免在编译时或运行时出现类型错误。因此，在《数据结构》教科书中，通常默认线性表中的元素类型是相同的。*/
```

# 树

```c++17
根节点：没有双亲节点的节点称为根节点
叶节点：没有子节点的节点称为叶节点
双亲节点：有子节点的节点称为双亲节点
子节点：有双亲节点的节点称为子节点
兄弟节点：同一双亲节点的子节点互称为兄弟节点

二叉树：每个结点最多有两个子结点的树称为二叉树。
完全二叉树：设深度为k的二叉树，除第k层外，其余各层的结点数都达到最大值，第k层所有的结点都连续集中在最左边，这样的二叉树称为完全二叉树。
满二叉树：每个结点要么是叶子结点，要么有两个子结点的二叉树称为满二叉树。

完全二叉树：    满二叉树：      普通二叉树：
    1               1              1
   / \             / \            / \
  2   3           2   3          2   3
 / \             / \ / \        /     \
4   5           4  5 6  7      4       5
```

# 查找

```c++17
查找方法
    1.顺序查找（线性查找）
        适用于无序表
        时间复杂度O(n)
        从头到尾逐个比较
    2.二分查找（折半查找）
        适用于有序表
        时间复杂度O(logn)
        每次将查找范围缩小一半
        需要随机访问能力
    3.分块查找（索引顺序查找）
        将表分成多个块，每个块有序
        先查找索引表，再在对应块内查找
        时间复杂度O(logn+n)
        适用于表较大，且查询频繁的情况
    4.树表查找
        基于树结构的查找方法
        包括二叉排序树、平衡树等
        时间复杂度O(logn)
        适用于频繁查询的情况
    5.哈希查找（）
        利用哈希函数将关键字映射到存储地址
        时间复杂度O(1)
        适用于查询频繁的情况
        要求哈希函数均匀分布

```

# 排序

```c++17
内排序方法
    1.交换排序
        1.1冒泡排序
        1.2快速排序
    2.选择排序
        2.1简单选择排序
        2.2堆排序
    3.插入排序
        3.1直接插入排序
        3.2折半插入排序
        3.3链表插入排序
        3.4希尔排序
    4.归并排序

    5.基数排序
```

```c++17
/*时间复杂度  空间复杂度  是否稳定*/

// 1.1
// 冒泡排序  O(n^2)  O(1)  稳定    
// 重复遍历数组，每次比较相邻元素，如果顺序错误就交换
void bubbleSort(int arr[], int n) {
    for (int i = 0; i < n-1; i++) {
        for (int j = 0; j < n-i-1; j++) {
            if (arr[j] > arr[j+1]) {
                // 交换
                int temp = arr[j];
                arr[j] = arr[j+1];
                arr[j+1] = temp;
            }
        }
    }
}

/**************************************************************/
// 首选qsort() 
// 原型：void qsort(void *base, size_t nitems, size_t size, int (*compar)(const void *, const void*))

// 1.2
// 快速排序  O(n^2) <平均O(nlogn)>  O(logn)  不稳定
// 选择一个基准元素，小的放左边，大的放右边，递归排序（分而治之）
void quickSort(int arr[], int left, int right) {
    if (left >= right) { // left和right分别是数组的起始和结束下标
        return;
    }

    int pivot = arr[left];
    int i = left, j = right;
    while (i < j) {
        while (i < j && arr[j] >= pivot) {
            j--;
        }
        while (i < j && arr[i] <= pivot) {
            i++;
        }
        // 交换
        int temp = arr[i];
        arr[i] = arr[j];
        arr[j] = temp;
    }
    // 交换基准元素
    arr[left] = arr[i];
    arr[i] = pivot;

    quickSort(arr, left, i-1);
    quickSort(arr, i+1, right);
}
/**************************************************************/

// 2.1
// 简单选择排序  O(n^2)  O(1)  不稳定    
// 每次从未排序部分选择最小元素放到已排序部分的末尾
void selectionSort(int arr[], int n) {
    for (int i = 0; i < n-1; i++) {
        int min_idx = i;
        for (int j = i+1; j < n; j++) {
            if (arr[j] < arr[min_idx]) {
                min_idx = j;
            }
        }
        // 交换
        int temp = arr[i];
        arr[i] = arr[min_idx];
        arr[min_idx] = temp;
    }
}

// 3.1
// 直接插入排序 O(n^2)  O(1)  稳定    
// 每次将一个未排序元素插入到已排序部分的合适位置
void insertionSort(int arr[], int n) {
    for (int i = 1; i < n; i++) {
        int key = arr[i];
        int j = i - 1;
        while (j >= 0 && arr[j] > key) {
            arr[j + 1] = arr[j];
            j--;
        }
        arr[j + 1] = key;
    }
}

// 3.2
// 折半（二分）插入排序 O(n^2）<比较次数O(nlogn），移动次数O(n^2>  O(1)  稳定  
// 使用二分查找找到插入位置，减少比较次数(优化版直接插入排序)
void binaryInsertionSort(int arr[], int n) {
    for (int i = 1; i < n; i++) {
        int key = arr[i];
        int left = 0, right = i - 1;
        while (left <= right) {
            int mid = left + (right - left) / 2;
            if (arr[mid] > key) {
                right = mid - 1;
            } else {
                left = mid + 1;
            }
        }
        for (int j = i - 1; j >= left; j--) {
            arr[j + 1] = arr[j];
        }
        arr[left] = key;
    }
}

// 3.3
// 链表插入排序 O(n^2)  O(1)  稳定    
// 将未排序元素逐个插入到已排序链表的适当位置(只需改变指针，不用移动元素是链表的优势)
void linkedListInsertionSort(linklist H) {
    if (H == NULL || H->next == NULL) {
        return -1;
    }

    linklist sorted = list_create();
    listnode* curr = H->next;
    while (curr != NULL) {
        listnode* next = curr->next;
        list_tail_insert(sorted, curr->data);
        curr = next;
    }
    H->next = sorted->next;
    free(sorted);
}
```
