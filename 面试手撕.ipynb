提供一个完整的C语言基础队列实现，包括：

    使用动态数组（环形缓冲区）

    支持泛型数据（通过 void*）

    包含基本操作：初始化、入队、出队、判空、判满、销毁

```c++17
#include <stdio.h>
#include <stdlib.h>
#include <stdbool.h>
#include <string.h>

// 队列结构体
typedef struct {
    void **data;        // 存储元素指针的数组
    int front;          // 队头索引
    int rear;           // 队尾索引
    int capacity;       // 队列容量
    int size;           // 当前元素个数
} Queue;

// 初始化队列
Queue* queue_create(int capacity) {
    Queue *q = (Queue*)malloc(sizeof(Queue));
    if (!q) return NULL;

    q->data = (void**)malloc(sizeof(void*) * capacity);
    if (!q->data) {
        free(q);
        return NULL;
    }

    q->front = 0;
    q->rear = 0;
    q->capacity = capacity;
    q->size = 0;
    return q;
}

// 销毁队列
void queue_destroy(Queue *q) {
    if (!q) return;
    free(q->data);
    free(q);
}

// 判空
bool queue_is_empty(Queue *q) {
    return q->size == 0;
}

// 判满
bool queue_is_full(Queue *q) {
    return q->size == q->capacity;
}

// 入队（返回 true 成功，false 失败）
bool enqueue(Queue *q, void *item) {
    if (!q || queue_is_full(q)) {
        return false;
    }

    q->data[q->rear] = item;
    q->rear = (q->rear + 1) % q->capacity;
    q->size++;
    return true;
}

// 出队（返回元素指针，失败返回 NULL）
void* dequeue(Queue *q) {
    if (!q || queue_is_empty(q)) {
        return NULL;
    }

    void *item = q->data[q->front];
    q->front = (q->front + 1) % q->capacity;
    q->size--;
    return item;
}

// 获取队头元素（不出队）
void* queue_front(Queue *q) {
    if (!q || queue_is_empty(q)) {
        return NULL;
    }
    return q->data[q->front];
}

// 获取队列元素个数
int queue_size(Queue *q) {
    return q->size;
}

// ========== 测试示例 ==========
int main() {
    // 创建一个容量为 5 的队列
    Queue *q = queue_create(5);
    if (!q) {
        printf("队列创建失败\n");
        return 1;
    }

    // 入队一些整数（用 int 举例）
    int a = 10, b = 20, c = 30, d = 40, e = 50;

    printf("入队: %d\n", a); enqueue(q, &a);
    printf("入队: %d\n", b); enqueue(q, &b);
    printf("入队: %d\n", c); enqueue(q, &c);
    printf("入队: %d\n", d); enqueue(q, &d);
    printf("入队: %d\n", e); enqueue(q, &e);

    // 尝试再入队一个（会失败，因为已满）
    int f = 60;
    if (!enqueue(q, &f)) {
        printf("入队失败：队列已满\n");
    }

    printf("队列大小: %d\n", queue_size(q));
    printf("队头元素: %d\n", *(int*)queue_front(q));

    // 出队所有元素
    printf("\n出队顺序: ");
    while (!queue_is_empty(q)) {
        int *val = (int*)dequeue(q);
        printf("%d ", *val);
    }
    printf("\n");

    // 销毁队列
    queue_destroy(q);
    return 0;
}
```

提供一个完整的C语言基础链表实现，包括：

    基本操作：创建节点、插入（头插/尾插）、删除和遍历功能

```c++17
#include <stdio.h>
#include <stdlib.h>

// 定义链表节点结构
typedef struct Node {
    int data;
    struct Node* next;
} Node;

// 创建新节点
Node* createNode(int data) {
    Node* newNode = (Node*)malloc(sizeof(Node));
    if (!newNode) {
        printf("内存分配失败\n");
        exit(1);
    }
    newNode->data = data;
    newNode->next = NULL;
    return newNode;
}

// 尾插法
void insertTail(Node** head, int data) {
    Node* newNode = createNode(data);
    if (*head == NULL) {
        *head = newNode;
        return;
    }
    Node* temp = *head;
    while (temp->next != NULL) {
        temp = temp->next;
    }
    temp->next = newNode;
}

// 头插法
void insertHead(Node** head, int data) {
    Node* newNode = createNode(data);
    newNode->next = *head;
    *head = newNode;
}

// 删除指定值的节点（删除第一个匹配项）
void deleteNode(Node** head, int key) {
    if (*head == NULL) return;

    Node* temp = *head;
    // 如果头节点就是要删除的
    if (temp->data == key) {
        *head = temp->next;
        free(temp);
        return;
    }

    // 查找要删除的节点
    Node* prev = NULL;
    while (temp != NULL && temp->data != key) {
        prev = temp;
        temp = temp->next;
    }

    if (temp == NULL) {
        printf("未找到值 %d\n", key);
        return;
    }

    prev->next = temp->next;
    free(temp);
}

// 遍历并打印链表
void printList(Node* head) {
    if (head == NULL) {
        printf("链表为空\n");
        return;
    }
    Node* temp = head;
    while (temp != NULL) {
        printf("%d -> ", temp->data);
        temp = temp->next;
    }
    printf("NULL\n");
}

// 释放整个链表
void freeList(Node** head) {
    Node* temp = *head;
    while (temp != NULL) {
        Node* next = temp->next;
        free(temp);
        temp = next;
    }
    *head = NULL;
}

// 获取链表长度
int getLength(Node* head) {
    int count = 0;
    Node* temp = head;
    while (temp != NULL) {
        count++;
        temp = temp->next;
    }
    return count;
}

// 测试代码
int main() {
    Node* head = NULL;

    printf("=== 尾插法插入 1,2,3 ===\n");
    insertTail(&head, 1);
    insertTail(&head, 2);
    insertTail(&head, 3);
    printList(head);  // 1 -> 2 -> 3 -> NULL

    printf("\n=== 头插法插入 0 ===\n");
    insertHead(&head, 0);
    printList(head);  // 0 -> 1 -> 2 -> 3 -> NULL

    printf("\n=== 删除值 2 ===\n");
    deleteNode(&head, 2);
    printList(head);  // 0 -> 1 -> 3 -> NULL

    printf("\n=== 删除值 5（不存在） ===\n");
    deleteNode(&head, 5);

    printf("\n=== 当前链表长度: %d ===\n", getLength(head));  // 3

    printf("\n=== 释放链表 ===\n");
    freeList(&head);
    printList(head);  // 链表为空

    return 0;
}
```

### 两数之和（哈希表）

给定一个整数数组 nums 和一个整数目标值 target，请你在该数组中找出 和为目标值 target  的那 两个 整数，并返回它们的数组下标。

```c++17
#include "uthash.h"

typedef struct hashTable_t {
    int key;
    int val;
    UT_hash_handle hh;
}Hash_table;

Hash_table* hashtable;

Hash_table* find(int ikey) {
    Hash_table* tmp;
    HASH_FIND_INT(hashtable, &ikey, tmp);
    return tmp;
}

void insert(int ikey, int ival) {
    Hash_table* it = find(ikey);
    if (it == NULL) {
        Hash_table* tmp = malloc(sizeof(Hash_table));
        tmp->key = ikey, tmp->val = ival;
        HASH_ADD_INT(hashtable, key, tmp);
    } else {
        it->val = ival;
    }
}

int* twoSum(int* nums, int numsSize, int target, int* returnSize) {
    hashtable = NULL;
    for (int i = 0; i < numsSize; i++) {
        Hash_table* it = find(target - nums[i]);
        if (it != NULL) {
            int* ret = malloc(sizeof(int) * 2);
            ret[0] = it->val, ret[1] = i;
            *returnSize = 2;
            return ret;
        }
        insert(nums[i], i);
    }
    *returnSize = 0;
    return NULL;
}
```

### 两个数组的交集

```c++17
#include <stdlib.h>

static int _cmp(const void* a, const void* b){
    return *(int* )a - *(int* )b;
}

int* intersection(int* nums1, int nums1Size, int* nums2, int nums2Size, int* returnSize) {
    int* returnedNums = (int*)malloc(1000 * sizeof(int));

    qsort(nums1, nums1Size, sizeof(int), _cmp);
    qsort(nums2, nums2Size, sizeof(int), _cmp);

    int index = 0;

    // 双指针
    int i,j = 0;

    while((i < nums1Size) && (j < nums2Size)){
        // 筛非重复,以相同的最后一个为准
        if(i > 0 && (nums1[i-1] == nums1[i])){
            i++; // 处理过了，+1看下一个重复不重复
            continue;
        }

        if(j > 0 && (nums2[j-1] == nums2[j])){
            j++;
            continue;
        }
        // 筛非重复,以相同的最后一个为准

        if(nums1[i] == nums2[j]){
            returnedNums[index] = nums1[i];
            index++;
            i++;
            j++;
        }else if(nums1[i] < nums2[j]){ // 小的nums1慢了
            i++;
        }else{// nums2慢了
            j++;
        }
    }

    *returnSize = index;

    return returnedNums;

}
```

### 交替合并字符串

<small>双指针</samll>

```c++17
char word1[] = "ace";  
char word2[] = "bdfgh"; 

char* mergeAlternately(char* word1, char* word2) {
    int len1 = strlen(word1);
    int len2 = strlen(word2);
    
    char* result = (char*)malloc(len1 + len2 + 1);
    if (!result) return NULL;
    
    int i = 0, j = 0, k = 0;
    
    while (i < len1 || j < len2) {
        if (i < len1) result[k++] = word1[i++];
        if (j < len2) result[k++] = word2[j++];
    }
    
    result[k] = '\0';
    return result;
}

char* result = mergeAlternately(word1, word2);
printf("%s\n", result);
free(result);//谁用谁负责释放内存
```

```c++17
#include <string.h>
#include <stdlib.h>
#include <stdio.h>

char word1[] = "ababababab";  
char word2[] = "abab"; 

//欧几里得迭代法
/*int gcd(int a, int b) {
    while (b != 0) {
        int temp = b;
        b = a % b;
        a = temp;
    }
    return a;
}*/

//欧几里得递归法
int gcd(int a, int b) {
    if (b == 0) {
        return a;
    }
    return gcd(b, a % b);
}

char* gcdOfStrings(char* str1, char* str2) {
    //如果两个字符串逆序拼接不相等，则没有最大公因子
    if (strcmp(strcat(str1, str2), strcat(str2, str1)) != 0) {
        static const char empty[] = "";
        return empty;

    }else{
        //辗转相除法（欧几里得算法）求最大公约字符长度gcd_len
        //取第一个字符串前gcd_len个字符即为最大公约字符串
        int gcd_len = gcd(strlen(str1), strlen(str2));
        char* result = (char*)malloc(gcd_len + 1);
        strncpy(result, str1, gcd_len);
        result[gcd_len] = '\0';

        return result;

    }
}

printf("%s", gcdOfStrings(word1, word2));
free(result);
```

### 大数求和 $\sum_{k=1}^{n} k!$

```c++17
// 用数组手写大数乘法/加法，求 1! + 2! + ... + 70!
// unsigned long long（最多到 20!）
// 大数表示：digits[j] 存第 j 位十进制数字，低位在前（digits[0] 是个位）

#define MAXD 256   // 足够容纳 70!（约 100 位）

int fact_d[MAXD] = {0};   // 当前阶乘 i!
int sum_d[MAXD]  = {0};   // 累加和
int f_len = 1, s_len = 1;
fact_d[0] = 1;            // 0! = 1

for (int i = 1; i <= 70; i++) {
    // ---- 大数乘法：fact = fact * i ----
    int carry = 0;
    for (int j = 0; j < f_len; j++) {
        int tmp = fact_d[j] * i + carry;
        fact_d[j] = tmp % 10;   // 保留个位
        carry     = tmp / 10;   // 进位给下一位
    }
    while (carry) {              // 进位让数字变长
        fact_d[f_len++] = carry % 10;
        carry /= 10;
    }

    // ---- 大数加法：sum += fact ----
    carry = 0;
    for (int j = 0; j < f_len; j++) {
        int tmp = sum_d[j] + fact_d[j] + carry;
        sum_d[j] = tmp % 10;
        carry    = tmp / 10;
    }
    int j = f_len;
    while (carry) {              // 超出 fact 长度的进位（不能覆盖 sum 原有高位）
        int tmp = sum_d[j] + carry;
        sum_d[j] = tmp % 10;
        carry    = tmp / 10;
        j++;
    }
    if (j > s_len) s_len = j;
}

// 高位在前打印结果
printf("1! + ... + 70! = ");
for (int i = s_len - 1; i >= 0; i--)
    printf("%d", sum_d[i]);

// 说明：
//  1) 复杂度 O(位数 × n)，70! 约 100 位，完全够用；
//  2) 想更高效可改成"万进制"（每个元素存 0~9999），位数只剩 1/4，逻辑相同；
//  3) 对比验证：该结果与 Python 的 math.factorial 求和一致：
//     12152211181772879322772500884098381612524494662396029015406215899272824694011444828896528920420940313
```
